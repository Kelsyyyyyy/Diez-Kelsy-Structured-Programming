# 1. Introduction

A function that calls itself is called a **recursive function**.

Recursion is based on the following idea:

> A complex problem can sometimes be solved by reducing it to a smaller version of the same problem.

This process continues until a simple case is reached.

## The Two Essential Components of Recursion

### Base Case
The condition that stops the recursion.

### Recursive Case
The part of the function that calls itself using a smaller version of the problem.

### General Form
```python
def function(problem):
    if simple_case:
        # base case
        do_something
    else:
        # recursive case
        function(smaller_problem)
```

# 2. First Recursive Program

````python
def countdown(n):
    if n == 0:
        print("Liftoff!")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
````

### Questions
1. Which statement makes this function recursive?

    *The statement `countdown(n - 1)` makes this function recursive, because the function calls itself with a smaller value.*
---

2. What is the base case?

    *The base case is `if n == 0`, which stops the recursion and prints "Liftoff!".*
---

3. Why does the function eventually stop?

    *The function eventually stops because each recursive call reduces `n` by 1. When `n` reaches 0, the base case is triggered and no more recursive calls are made.*
---

4. What value is passed during each new function call?

    *Each new call receives the value `n - 1`, which is one less than the previous call.*
---

# 3. Understanding Function Calls

When a recursive function calls itself, Python temporarily pauses the current function call and starts executing the new one.

After the base case is reached, the function calls finish one by one in reverse order. This behavior is known as the **call stack**.

````python
def explore(level):
    print("Entering level", level)

    if level > 1:
        explore(level - 1)

    print("Leaving level", level)

explore(3)
````

### Before running the program
Predicted output:
```
Entering level 3
Entering level 2
Entering level 1
Leaving level 1
Leaving level 2
Leaving level 3
```

### After running the program
Actual output:
```
Entering level 3
Entering level 2
Entering level 1
Leaving level 1
Leaving level 2
Leaving level 3
```

### Questions
1. Which message appears first?

    *"Entering level 3" appears first, because that is the initial call made to the function.*

2. Which message appears last?

    *"Leaving level 3" appears last, because the first call to `explore` is the last one to finish after all recursive calls have completed.*

3. Why are the 'Leaving' messages printed in reverse order?

    *The 'Leaving' messages are printed in reverse order because of how the call stack works. Each function call is paused while the next recursive call runs. Once the deepest call (level 1) finishes, the calls return in reverse order — level 1 finishes first, then level 2, then level 3.*

# 5. Returning Values Recursively

````python
def treasures(level):
    if level == 0:
        return 0
    else:
        return 1 + treasures(level - 1)

treasures(4)
````

Trace the execution:

```text
treasures(4)
= 1 + treasures(3)
= 1 + 1 + treasures(2)
= 1 + 1 + 1 + treasures(1)
= 1 + 1 + 1 + 1 + treasures(0)
= 1 + 1 + 1 + 1 + 0
= 4
```

# 6. Recursive Summation

````python
def coins(level):
    if level == 0:
        return 0
    else:
        return level + coins(level - 1)

coins(5)
````

The value returned by `coins(5)` is **15**. Here is every step of the recursive process:

```text
coins(5)
= 5 + coins(4)
= 5 + 4 + coins(3)
= 5 + 4 + 3 + coins(2)
= 5 + 4 + 3 + 2 + coins(1)
= 5 + 4 + 3 + 2 + 1 + coins(0)
= 5 + 4 + 3 + 2 + 1 + 0
= 15
```

# 7. Writing a Recursive Function

````python
def countdown(n):
    if n == 0:
        print("Done!")
    else:
        print(n)
        countdown(n - 1)

countdown(4)
````

# 8. Common Errors in Recursive Programming

````python
# This function has a bug — do NOT run it as-is (it will crash)
# def mystery(n):
#     print(n)
#     mystery(n - 1)
#
# mystery(5)

# Corrected version:
def mystery(n):
    if n == 0:          # base case added
        return
    print(n)
    mystery(n - 1)

mystery(5)
````

### Questions
1. Does the function stop?

    *No, the original function never stops. It keeps calling itself indefinitely because there is no base case to end the recursion.*
---
2. What error is eventually generated?

    *Python raises a `RecursionError: maximum recursion depth exceeded` error.*
---
3. Why does this error occur?

    *This error occurs because the function has no base case. Without a stopping condition, the function calls itself forever, filling up the call stack until Python's recursion limit is reached.*
---
4. How can the function be corrected?

    *A base case must be added. For example, adding `if n == 0: return` stops the recursion when `n` reaches 0, preventing the infinite loop.*
---

# 9. Recursion versus Iteration
